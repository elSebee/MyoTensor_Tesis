    \item \textbf{Comparación Anatómica y Validación Estadística Formal.} Protocolo de contrastes pareados de medidas repetidas ($N = 10$, $\alpha = 0.05$) para corroborar formalmente el impacto de la ubicación muscular (Extensor Digitorum frente a Flexor Digitorum Superficialis) y contrastar la hipótesis central de superioridad algorítmica (Deep Learning vs. Machine Learning clásico). Finalmente, los hallazgos se integran en una matriz de decisión multicriterio de ingeniería (AHP ponderado) para definir la topología muscular y el clasificador óptimo a transferir al microcontrolador ESP32-S3.

    \paragraph{Validación Estadística Formal de Hipótesis}
Para evaluar con rigor metodológico el impacto del sitio de registro y la hipótesis central del proyecto con un nivel de significancia $\alpha = 0.05$, se consolidaron las métricas pareadas ($F_1$-Score macro) obtenidas por los 10 participantes ($N = 10$). El protocolo se estructuró en tres ejes:
\begin{enumerate}
    \item \textbf{Impacto de la Ubicación Muscular:} Evaluación pareada del desempeño obtenido en el músculo Extensor Digitorum (ED) frente al Flexor Digitorum Superficialis (FDS) para determinar si la topología muscular influye de forma significativa en la exactitud y estabilidad del sistema.
    \item \textbf{Contraste de la Hipótesis Central (Deep Learning vs. ML Clásico):} Prueba pareada entre los clasificadores de Deep Learning (CNN-LSTM, CNN-TCN) y los algoritmos clásicos (SVM, Random Forest), evaluando si las redes neuronales presentan una ventaja estadísticamente significativa tanto en extensores como en flexores y a nivel global.
    \item \textbf{Matriz de Decisión Multicriterio de Ingeniería (AHP):} Integración cuantitativa de exactitud ($F_1$-Macro, $40\%$), latencia de control ($25\%$), memoria SRAM dinámica ($20\%$) y memoria Flash / soporte OTA ($15\%$) para seleccionar la arquitectura definitiva desplegable en el microcontrolador ESP32-S3.
\end{enumerate}

# Diagramas de Redes Neuronales con PlotNeuralNet (TikZ / LaTeX)

## 1. Entendimiento y Principios de Diseño en PlotNeuralNet

El repositorio **PlotNeuralNet** (HarisIqbal88) permite diagramar arquitecturas de redes neuronales en 3D utilizando el paquete **TikZ** de LaTeX. Se basa en una biblioteca de macros ubicada en `./layers/`:

* **Inclusión de Bioseñal Externa (`\node[canvas is zy plane]`)**: Mediante la librería `3d` de TikZ, la imagen física de la señal mioeléctrica (`semg.jpeg`) se proyecta en el plano isométrico $Y-Z$ previo a la red, conectándose mediante un vector direccional al tensor de entrada digital.
* **`Box`**: Dibuja paralelepípedos 3D definidos por tres dimensiones escalables:
  * **`depth` y `height`**: Representan la dimensión temporal de la serie de tiempo sEMG (longitud de ventana $W=250$, submuestreo tras pooling $125 \rightarrow 62 \rightarrow 1$).
  * **`width`**: Representa la profundidad en canales/filtros de características ($1 \rightarrow 32 \rightarrow 64$ filtros o unidades densas).
  * Etiquetas: `xlabel` (canales/filtros), `zlabel` (dimensión temporal/secuencia) y `caption` (nombre de la capa bajo el bloque).
* **`RightBandedBox`**: Extiende `Box` agregando una banda de color en su cara derecha. Es la representación canónica para capas combinadas como **Convolución + Activación (ReLU)** o **Convolución + BatchNormalization**.
* **`Ball`**: Dibuja una esfera que denota operaciones tensoriales puntuales, como la **suma elemento a elemento** ($\oplus$, `logo=+$`) en conexiones residuales (*skip connections*).
* **Puntos de anclaje y conexiones**: Cada bloque expone coordenadas (`-west`, `-east`, `-north`, `-south`). Las conexiones feedforward viajan de cara este a cara oeste mediante `\draw [connection] (capa1-east) -- node {\midarrow} (capa2-west);`. Las conexiones residuales se desvían ortogonalmente por la parte inferior (`-south`) para conectarse a las esferas de adición.

## 3. Red 2: CNN-TCN (`CNN_TCN_MyoTensor`)

### Especificación de Arquitectura y Tensores
* **Bioseñal Cruda**: Imagen proyectada en el plano $Y-Z$ (`semg.jpeg`).
* **Entrada**: Ventana de sEMG de $250\text{ ms}$ a $1\text{ kHz}$ ($250 \times 1$).
* **Proyección Inicial**:
  * `Conv1D`: 32 filtros, kernel $K=1$, padding *same* $\rightarrow (250 \times 32)$.
* **Bloque Residual TCN 1 (Dilatación $d=1$)**:
  * 2 Convoluciones causales dilatadas ($K=3, d=1$, 32 filtros, ReLU) + BatchNormalization + SpatialDropout (0.2).
  * Conexión residual directa 1x1 (identidad).
  * Suma elemento a elemento ($\oplus$) + ReLU $\rightarrow (250 \times 32)$.
* **Bloque Residual TCN 2 (Dilatación $d=2$)**:
  * 2 Convoluciones causales dilatadas ($K=3, d=2$, 32 filtros, ReLU) + BatchNormalization + SpatialDropout (0.2). Campo receptivo efectivo expandido.
  * Conexión residual directa.
  * Suma elemento a elemento ($\oplus$) + ReLU $\rightarrow (250 \times 32)$.
* **Bloque Residual TCN 3 (Dilatación $d=4$)**:
  * 2 Convoluciones causales dilatadas ($K=3, d=4$, 32 filtros, ReLU) + BatchNormalization + SpatialDropout (0.2). Cobertura completa de la ventana temporal de 250 ms.
  * Conexión residual directa.
  * Suma elemento a elemento ($\oplus$) + ReLU $\rightarrow (250 \times 32)$.
* **Global Average Pooling 1D (GAP)**:
  * Colapso de la dimensión temporal completa (250 instantes) a un vector plano de 32 dimensiones $\rightarrow (32)$.
* **Capa Densa Intermedia**:
  * `Dense`: 32 unidades, activación ReLU $\rightarrow (32)$.
* **Capa de Salida**:

### Gráfico 1: Ganancia Marginal por Componente (EXP01)


````python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from pathlib import Path

# Configuración de estilo global
sns.set_theme(style="whitegrid", font="sans-serif")
plt.rcParams.update({
    "font.family": "sans-serif",
    "font.size": 10.5,
    "axes.labelsize": 11.5,
    "axes.titlesize": 12.5,
    "xtick.labelsize": 10.0,
    "ytick.labelsize": 10.0,
    "legend.fontsize": 9.2,
    "figure.titlesize": 13.5,
})

MODEL_COLORS = {
    "RF": "#009E73",       # Emerald Teal
    "SVM":           "#E69F00",       # Amber Orange
    "CNN-LSTM":      "#4C72B0",       # Steel Blue
    "CNN-TCN":       "#9467BD",       # Royal Purple
}

# Datos extraídos de Fase 1 (Ganancia Marginal en F1-Score Macro vs Baseline Crudo)
data_exp01 = {
    "FDS": {
        "RF":       {"Steady-State": 7.09, "Normalización MVC": -0.74, "Ganancia Neta": 6.34},
        "SVM":      {"Steady-State": 0.86, "Normalización MVC": -0.93, "Ganancia Neta": -0.07},
        "CNN-LSTM": {"Steady-State": 7.25, "Normalización MVC":  0.09, "Ganancia Neta": 7.34},
        "CNN-TCN":  {"Steady-State": 4.37, "Normalización MVC": -10.27, "Ganancia Neta": -5.89}
    },
    "ED": {
        "RF":       {"Steady-State": 3.70, "Normalización MVC": 0.34,  "Ganancia Neta": 4.04},
        "SVM":      {"Steady-State": 3.78, "Normalización MVC": 0.80,  "Ganancia Neta": 4.59},
        "CNN-LSTM": {"Steady-State": 2.45, "Normalización MVC": 0.03,  "Ganancia Neta": 2.48},
        "CNN-TCN":  {"Steady-State": 1.65, "Normalización MVC": -0.90, "Ganancia Neta": 0.75}
    }
}

models = ["RF", "SVM", "CNN-LSTM", "CNN-TCN"]
components = ["Steady-State", "Normalización MVC", "Ganancia Neta"]
n_models = len(models)
bar_width = 0.19

fig, axes = plt.subplots(1, 2, figsize=(10.0, 4.3), sharey=True)

for idx, (muscle, title) in enumerate([("FDS", "A. Flexor Digitorum Superficialis (FDS)"),
                                       ("ED", "B. Extensor Digitorum (ED)")]):
    ax = axes[idx]
    x_base = np.arange(len(components))
    
    for m_idx, model in enumerate(models):
        offset = (m_idx - (n_models - 1) / 2) * bar_width
        vals = [data_exp01[muscle][model][c] for c in components]
        bars = ax.bar(x_base + offset, vals, width=bar_width * 0.90,
                      label=model if idx == 0 else "",
                      color=MODEL_COLORS[model], edgecolor="black", linewidth=0.7, alpha=0.9)
        
        for bar, val in zip(bars, vals):
            y_pos = bar.get_height()
            va = "bottom" if y_pos >= 0 else "top"
            stagger = 2.0 if (m_idx % 2 == 1) else 0.0
            offset_pts = (2.2 + stagger) if y_pos >= 0 else (-2.2 - stagger)
            ax.annotate(f"{val:+.1f}",
                        (bar.get_x() + bar.get_width() / 2, y_pos),
                        ha="center", va=va, fontsize=7.8, fontweight="bold",
                        xytext=(0, offset_pts), textcoords="offset points")

    ax.axhline(0, color="black", linestyle="-", linewidth=0.8, alpha=0.6)
    ax.set_title(title, fontweight="bold", fontsize=11.5)
    ax.set_xticks(x_base)
    ax.set_xticklabels(components, fontweight="bold", fontsize=9.5)
    ax.set_ylim(-13.5, 11.5)
    if idx == 0:
        ax.set_ylabel("Ganancia en F1 (Puntos Porcentuales)", fontweight="bold", fontsize=10.5)
        ax.legend(title="Arquitectura", loc="lower left", frameon=True, fontsize=9.0, title_fontsize=9.5)

plt.suptitle("Ganancia Marginal en F1-Score por Componente de Preprocesamiento", fontweight="bold", fontsize=12.5, y=0.98)
plt.tight_layout()

save_path = Path("../../src/experimentos/logs/exp01.png")
save_path.parent.mkdir(parents=True, exist_ok=True)
fig.savefig(save_path, dpi=300, bbox_inches="tight")
plt.show()
````
